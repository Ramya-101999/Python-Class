```python
# Batch processing with glob.glob("*.csv")

import glob
import pandas as pd

all_dfs = []
for filepath in glob.glob("*.csv"):
    df = pd.read_csv(filepath,low_memory=False)
    df["source_file"] = filepath
    all_dfs.append(df)

new = pd.concat(all_dfs,ignore_index=True)
print(new)
print(all_dfs)
```

```python
# WebScraping with BeautifulSoup (BS4)

# syntax - BeautifulSoup(html_text,"html.parser")

# Scrapping a real website 

import requests
from bs4 import BeautifulSoup

resp = requests.get("https://pypi.org/project/pandas/")
print(resp.status_code)
soup = BeautifulSoup(resp.text,"html.parser")
print(soup)
```

```python
print(soup.find("title").get_text())
print(soup.find("meta",property="og:title")["content"])
print(soup.find("meta",attrs={"name":"description"})["content"])
```

| Method                        | What it does                                          |
|------------------------------------|--------------------------------------------------------------|
| soup.find(tag, class_=...)              | Finds the FIRST matching element                                 |
| soup.find_all(tag, class_=...)            | Finds ALL matching elements, returned as a list                    |
| soup.find(tag, attrs={"name":...})          | Finds a tag by a specific attribute value (e.g. a meta tag's "name") |
| soup.select("css selector")                   | Finds elements using CSS-selector syntax                              |
| element.get_text(strip=True)                    | Extracts visible text, with whitespace trimmed                          |
| element.get("href") / element["href"]              | Extracts a specific attribute's value from a tag                          |
