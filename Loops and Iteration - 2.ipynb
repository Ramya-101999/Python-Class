# List Comphrehensions

```python
# This is Ternary Operator
# syntax -
# variable = value_if_true if condition  else value_if_else
i = input("Enter d for deliverd and c for cancelled")
s = "delivered" if i=="d" else "cancelled"
print(s) 

age = 90
a = 30 if age>18 else 70
print(a)
```

```python
# List comphrehensions
# A list comphrehension is a compact way to build a new list from an existing one, in a single line

s = [2,3,4,5,6]
p = []
for i in s:
    k = i**2
    p.append(k)
print(p)

# alternate method - list comphrehension

# syntax --->  x = [y*2  for y in numbers]

s = [2,3,4,5,6]
p = [i**2 for i in s]
print(p)
```

```python
# Add elements into p after converting every string in s in uppercase using list comphrehension:

s = ["hi","hello","bottle","keyboard","Mouse"]
p = []

for i in s:
    b = i.upper()
    p.append(b)
print(p)

# list comphrehension method
s = ["hi","hello","bottle","keyboard","Mouse"]
p = [i.upper() for i in s]
print(p)
```

```python
# convert each to fahrenheit using (c*9/5)+32 using list comphrehension
temps = [0,20,30,40]
t = []

t = [(c*9/5)+32 for c in temps]
print(t)
```

```python
# Filter with list comphrehension

# n*2 only if n is even

a = [7,9,10,56,78,32,33]
l = []

l = [n*2 for n in a if n%2==0]
print(l)
```

```python
# add words into the list m from list t only if they have more than 3 letters

t = ["cat","elephant","mat","bat","tim","time","bottle"]
m = []

m = [w for w in t if len(w)>3]
print(m)
```

```python
# capitalize the word only if word starts with t


t = ["cat","elephant","mat","bat","tim","time","bottle","tounge"]
m = []

m = [w.title() for w in t if w.startswith("t")]
print(m)
```

```python
# Enter the length of the word for every word in the list

t = ["cat","elephant","mat","bat","tim","time","bottle","tounge"]
m =[]

m = [len(w) for w in t]
print(m)
```

```python
# different values for every item (values depends on a condition)

a = [78,90,56,40,34,77,23]
grade = ["Pass" if t>40 else "Fail" for t in a]
print(grade)
```

```python
# Enter small if the word's length is less than 5 else big

t = ["cat","elephant","mat","bat","tim","time","bottle","tounge"]
words = ["Small" if len(w) < 5 else "Big" for w in t]
print(words)
```

```python
# Enter success if delivered otherwise enter fail

statuses = ["cancelled","pending","closed","cancelled","pending","delivered","delivered"]

s = ["Success" if status == "delivered" else "Fail" for status in statuses ]
print(s)
```

| Pattern                          | Syntax                                    |
|--------------------------------------|--------------------------------------------|
| Transform every item                    | [expression for item in list]                |
| Filter only                              | [item for item in list if condition]           |
| Transform + filter                        | [expression for item in list if condition]       |
| Value depends on condition (no filtering)  | [val_if_true if condition else val_if_false for item in list] |

```python
# With range()

# get square of every alternate number between 1-10

l1 = [i**2 for i in range(1,11,2)]
print(l1)
```

```python
uni = {"c1","c2","c3","c4","c2","c1"}
tag = ['Customer-c1']

tag = [f"Customer-{i}" for i in uni]
print(tag)
```

```python
prices = {"Pen":20,"Notebook":60,"Bag":800}
p = []

p = [k for k in prices.keys()]
print(p)
p = [v for v in prices.values()]
print(p)

p = [[k for k in prices] + [v for v in prices.values()]]
print(p)

# nested for loop in list comphrehension
p = [i for item in prices.items() for i in item]
print(p)
```

# Dictionary Comphrehension

```python
words = ["apple","banana","kiwi"]
word_lengths = {w:len(w) for w in words}
print(word_lengths)
```

```python
numbers = [78,99,200,100]
number_lengths = {n:len(str(n)) for n in numbers}
print(number_lengths)
```

```python
prices = {"Pen":20,"Notebook":60,"Bag":800}
p = {k:v for k,v in prices.items() if v>50}
print(p)
```

```python
celsius = {"Mon":30,"Tues":25,"Wed":32}
fahrenhiet = {day:(c*9/5)+32 for day,c in celsius.items()}
print(fahrenhiet)
```
