# Tuples 

A tuple is like a list, but locked
once created you cannot change,add or remove items

```python
t1 = (1,4.5,True,"Hello")
print(t1)
print(type(t1))
```

# Tuple Packing And Tuple Unpacking

```python
t2 = 3,4,5,6,14
print(type(t2))
print(t2)

#Tuple Unpacking
t_int,t_float,t_bool,t_string = t1
print(t_int)
print(t_float)
print(t_bool)
print(t_string)
```

```python
# tuple creation through tuple function/typecasting
l1 = tuple([1,2,3,4,5])
print(type(l1))

# Creating a tuple of single element
t = (1)
t1 = (1,)
print(type(t))
print(type(t1))

# tuple repetition
t_rep = (1,)*3
print(t_rep)

# list repetition
l_rep = [1]*3
print(l_rep)
```

```python
t_unpack = (1,2,3,4,5,6,6,7)
t_1,t_2,t_3,t_4,t_5,t_6,t_7,t_8 = t_unpack
print(t_unpack)

#  * can store multiple elements
*t1,t_2 = t_unpack
print(t_1)
print(t_2)

t_1,*t_2 = t_unpack
print(t_1)
print(t_2)

t_1,*t_2,t_3 = t_unpack
print(t_1)
print(t_2)
print(t_3)
```

# Tuple methods  
.count - count the numbers of values in a tuple 
.index - returns the first occurence

```python
t1 = (1,2,3,4,5,2,3,3,2)
print(t1.count(2))

print(t1.index(2))

print(len(t1))
print(sum(t1))
print(min(t1))
print(max(t1))
```

# Sets

sets - mutable, unordered, Heterogenous, No indexing, removes duplicate elements
(list,dictionary,set) can't be a member of set
Elements in the set must be unique and immutable

```python
s1 = (1,2,3,4,5)
print(s1)
print(type(s1))

s2 = {1}
print(type(s2))

s3 = {}     # cannot use to create empty set
print(type(s3))

s3 = set()
print(type(s3))
```
