# Tuples 

A tuple is like a list, but locked
once created you cannot change,add or remove items

```python
t1 = (1,4.5,True,"Hello")
print(t1)
print(type(t1))
```

# Tuple Packing And Tuple Unpacking

```python
t2 = 3,4,5,6,14
print(type(t2))
print(t2)

#Tuple Unpacking
t_int,t_float,t_bool,t_string = t1
print(t_int)
print(t_float)
print(t_bool)
print(t_string)
```

```python
# tuple creation through tuple function/typecasting
l1 = tuple([1,2,3,4,5])
print(type(l1))

# Creating a tuple of single element
t = (1)
t1 = (1,)
print(type(t))
print(type(t1))

# tuple repetition
t_rep = (1,)*3
print(t_rep)

# list repetition
l_rep = [1]*3
print(l_rep)
```

```python
t_unpack = (1,2,3,4,5,6,6,7)
t_1,t_2,t_3,t_4,t_5,t_6,t_7,t_8 = t_unpack
print(t_unpack)

#  * can store multiple elements
*t1,t_2 = t_unpack
print(t_1)
print(t_2)

t_1,*t_2 = t_unpack
print(t_1)
print(t_2)

t_1,*t_2,t_3 = t_unpack
print(t_1)
print(t_2)
print(t_3)
```

# Sets

sets - mutable, unordered, Heterogenous, No indexing, removes duplicate elements
(list,dictionary,set) can't be a member of set
Elements in the set must be unique and immutable
Set uses hashing method to store elements

```python
s1 = (1,2,3,4,5)
print(s1)
print(type(s1))

s2 = {1}
print(type(s2))

s3 = {}     # cannot use to create empty set
print(type(s3))

s3 = set()
print(type(s3))
```

```python
s1 = {91,88,2,3,4,5,5,4,90,100,3.4,"Hello",True,(4,5,6), (4,5,6,6)}
print(s1)
print(type(s1))
```

# hashability 

Hash - a unique number or fingerprint python generates randomly for every item stored inside the set, this is to store and look up items in a set. only immutable items can be hashed.

```python
print(hash("Hello"))
print(hash(5))
print(hash((1,2,3)))
print(hash(5.6))
```

# Set Operations
| Operation           | Method                        | Meaning                              |
|----------------------|----------------------------------|------------------------------------------|
| Union                | a.union(b) or a \| b            | Everything in either set                 |
| Intersection         | a.intersection(b) or a & b       | Only what's in BOTH sets                 |
| Difference           | a.difference(b) or a - b         | In a, but NOT in b                        |
| Symmetric Difference | a.symmetric_difference(b) or a ^ b | In one set OR the other, but not both  |

```python
set1= {"Mango","orange","Blueberry","Banana","cheese"}
set2 = {"Milk","pear","Tomato","cheese","Blueberry"}

print(set1.union(set2))
print(set1.intersection(set2))
print(set2.difference(set1))
print(set2.symmetric_difference(set1))
```

```python
# removing duplicates by converting into a set
l1  = [1,2,3,2,23,56,56,45]
s1 = set(l1)
print(s1)
```

# Set Methods

| Method/Operation                   | What it does                                     |
|-------------------------------------|---------------------------------------------------------|
| .add(x)                            | Adds one item                                              |
| .update(iterable)                  | Merges another set/list in, in place                        |
| .remove(x)                         | Removes x — errors if not found                              |
| .discard(x)                        | Removes x — safe, no error if missing                         |
| .pop()                             | Removes and returns an arbitrary item                          |
| .clear()                           | Empties the set completely                                      |
| .union(other) / \|                 | Combines both sets                                                |
| .intersection(other) / &           | Only items in BOTH sets                                            |
| .difference(other) / -             | In this set but not the other                                       |
| .symmetric_difference(other) / ^   | In one set or the other, but not both                                |
| .issubset(other)                   | True if all of this set's items are in other                          |
| .issuperset(other)                 | True if this set contains all of other's items                         |
| .isdisjoint(other)                 | True if the two sets share nothing in common                            |

```python
# add
s1 = {1,2,3,4}
s1.add(5)
print(s1)

# update
s1.update({6,7})
print(s1)
s1.update({4,5,6})
print(s1)

# update
s1.update({6,7})
print(s1)
s1.update({4,5,6})
print(s1)
s1.update([8,9,10])
print(s1)
s1.update((11,12))
print(s1)
s1.update((13,14,12,(1,2)))
print(s1)
```

```python
# remove
s1.remove(4)
print(s1)
  
s1.remove(3)
print(s1)

s1.remove(3)     #throws error as an element is not present in the set now
print(s1)
```

```python
# discard - safe method to remove elements because it does not throw error if element not present
s1.discard(5)
print(s1)
```

```python
# pop - pop is to remove any random item
s1.pop()
print(s1)
```

```python
# difference_update  - set operation to remove multiple items at once
s1 = {1,2,3,4,5}
s2 = {4,5}

s1.difference_update(s2)
print(s1)
```

```python
# issubset, issuperset, isdisjoint
s1 = {1,2,3,4,5}
s2 = {4,5}

print(s2.issubset(s1))
print(s1.issuperset(s2))
print(s1.isdisjoint(s2))
```

# Dictionary

Items are Stored as values in key:value:pair
Items are ordered, changable, indexing is done through keys, not allow duplicates (it should be unique)
keys must be immutable type, dictionary can store almost any datatype
Any type of value can be store as dictionary in python
Given key should appear only once
duplicate values will be overwrite existing values

```python
d1 = {}  # empty dictionary
d1 = {"Apple":"Red","Banana":"Yellow","Pear":"Green"}
print(d1["Apple"])

d1["Apple"] = "Blue"
print(d1)

d1["Orange"] = "Orange"
print(d1)

a = dict() # empty dictionary using dict method
print(a)

# converting 2 lists into dictionary
l1 = ["Blue", "Green","Orange"]
l2 = ["Mouse", "Laptop", "keyboard"]

# syntax - dict(zip(keys,values))
d2 = dict(zip(l1,l2))
print(d2)

# print(d2["Yellow"]) - will throw an key error when value is not present in the dictionary

# safe way to access a value -  .get("value to find, return value when not found") method
print(d2.get("Yellow","value not found"))
```

```python
# Dictionary Methods
# keys, values, items

print(d2.keys())
print(d2.values())
print(d2.items())
```

Methods for dictionary -
| Method | What it does |
|---|---|
| .get() | Safe access by key — no error if missing |
| .keys() | Returns all keys |
| .values() | Returns all values |
| .items() | Returns all key-value pairs |
| .update() | Updates one or more keys at once |
| .pop() | Removes a key and returns its value |
| .copy() | Returns a copy of the dictionary |
| .clear() | Empties the dictionary |

Return types for methods- 

| Method | Returns |
|---|---|
| .get() | Value, None, or default |
| .keys() | dict_keys view |
| .values() | dict_values view |
| .items() | dict_items view |
| .update() | None |
| .pop() | Removed value |
| .copy() | New dictionary |
| .clear() | None |

```python
# update
print(d1)
print(d1.update({"Rusk":"Brown", "samsung": "white"}))
print(d1)

# pop
d1.pop("Rusk")
print(d1)

# copy
d2 = d1.copy()
print(d2)

# popitem() - removes the last entered item
d2.popitem()
print(d2)

# setdefault() - adds the values only if missing in the dictionary
d2.setdefault("samsung","White")
print(d2)
```

```python
# dict from keys - to craete a dict from a list
# syntax - dict.fromkeys(list(keys),default value)
d1 = dict.fromkeys([1,2,3,4],0)
print(d1)

# operator to merge dictionaries
d1 = {"Apple":"Red","Banana":"Yellow"}
d2 = {"Mouse":"pink","keyboard":"Black"}
d3 = d1 | d2
print(d3)

d1 |= d2
print(d1)
```
