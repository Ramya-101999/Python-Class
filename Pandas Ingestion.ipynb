# Series Vs Dataframe

# 1. Series

```python
import pandas as pd

# Series Vs Dataframe

# Series
# syntax - 
# pd.series(data,index=None,name=None,dtype=None)

s = pd.Series([250,480,90,610], index = ["Mon","Tue","Wed","Thur"], name = "Daily Sales", dtype = float)
print(s)
print(type(s))
print(s.index)
print(s.values)

# A dictionary's keys become the index automatically
l = pd.Series({"Mon":250,"Tue":480,"Wed":90})
print(l)

# A scalar value broadcasts to every index label
h = pd.Series(5,index=["a","b","c"])
print(h)
```

| Parameter | Used for                                              |
|---------------|--------------------------------------------------------------|
| data              | The values — a list, dict, or a single scalar (broadcasts)      |
| index               | Custom labels for each value (defaults to 0,1,2...)                |
| name                  | A label for the Series itself (becomes the column name later)          |
| dtype                    | Force a specific data type                                                |

```python
# DataFrame - 2D Table

# syntax - 
# pd.DataFrame(data,inex=None,columns=None,dtype=None)
```

| Parameter | Used for                                              |
|---------------|--------------------------------------------------------------|
| data              | A dict of lists (columns), list of dicts (rows), or list of lists |
| index               | Custom row labels (defaults to 0,1,2...)                            |
| columns               | Column names — REQUIRED if data is a list of lists                    |
| dtype                    | Force a specific data type for all columns                              |

# 2. DataFrame

```python
# Creating a dataframe - 

# Dict of lists - each key becomes a column (this is the most common shape of dataframe)

df = pd.DataFrame({"day":["Mon","Tue","Wed","Thur"],"sales":[250,480,90,610]})
print(df)
print(type(df))
print(type(df["day"]))
print(type(df["sales"]))
```

```python
# list of dicts - each dict becomes one row

df1 = pd.DataFrame([{"name":"Ananya","score":88}, {"name":"Rahul","score":92}])
print(df1)
```

```python
# list of lists/ list of tuples - needs column parameter

df2 = pd.DataFrame([["Ananya",88],["Rahul",92]],columns=["name","score"])
print(df2)


df3 = pd.DataFrame([("Ananya",88),("Rahul",92)],columns=["name","score"])
print(df3)
```

```python
df4 = pd.DataFrame({"score":[88,92]}, index=["Ananya","Rahul"])
print(df4)
```

```python
# from a numpy array - each array row becomes a table row

import numpy as np
arr = np.array([[1,2,3],[4,5,6]])
df4 = pd.DataFrame(arr,columns=["col1","col2","col3"])
print(df4)
```

```python
# using zip

names = ["Ananya","Rahul"]
scores = [88,92]
df5 = pd.DataFrame(list(zip(names,scores)),columns = ["name","score"])
print(df5)


# a dict of series

s1 = pd.Series([10,20,30])
s2 = pd.Series([100,200,300])
df6 = pd.DataFrame({"col1":s1,"col2":s2})
print(df6)
```

```python
# Starting with an empty dataframe and adding columns

df7 = pd.DataFrame()
df7["name"] = ["Ananya","Rahul"]
df7["score"] = [88,92]
print(df7)

# Example:

d1 = {"name":1,"value":2}
d1["name"] = 4
print(d1)
d1["score"] = 92
print(d1)
```

```python
# Reading the anatomy - index,columns,values

df = pd.DataFrame({
    "day":["Mon","Tue","Wed","Thur"],
    "sales":[250,480,90,610],
    "region":["North","South","North","East"]
})

print(df)
print(df.index)   # index labels the rows
print(df.columns)  # columns labels the columns
print(df.values)  # values is the raw underlying data as a 2D numpy array - no labels just the data
```

| Parameter    | Used for                                                      |
|------------------|----------------------------------------------------------------------|
| filepath              | Path to the CSV file (required)                                          |

| na_values                             | Extra text values to treat as missing (e.g. ["N/A","unknown"])                               |
| parse_dates                             | List of column names to convert into real datetime objects                                     |

## Date Components

| Code | Meaning                          | Example (for 5 March 2026) |
|------|-----------------------------------|-----------------------------|
| %d   | Day of month (2-digit, zero-padded) | 05                          |
| %m   | Month (2-digit, zero-padded)        | 03                          |
| %y   | Year, 2-digit                        | 26                          |
| %Y   | Year, 4-digit                         | 2026                        |
| %j   | Day of the year (001–366)               | 064                         |
| %A   | Full weekday name                         | Thursday                    |
| %a   | Abbreviated weekday name                    | Thu                          |
| %B   | Full month name                               | March                          |
| %b   | Abbreviated month name                          | Mar                              |
| %U   | Week number of year (Sunday as first day)         | 09                                  |
| %W   | Week number of year (Monday as first day)           | 09                                    |
| %w   | Weekday as a number (0=Sunday, 6=Saturday)            | 4                                        |

## Timezone

| Code | Meaning                          | Notes                          |
|------|-----------------------------------|----------------------------------|
| %Z   | Timezone name                       | Empty if the datetime is naive (no timezone attached) |
| %z   | UTC offset (e.g. +0530)               | Empty if the datetime is naive                            |

## Escape

| Code | Meaning         |
|------|-------------------|
| %%   | A literal % character |
| Format string          | Matches an example like    |
|-----------------------------|---------------------------------|
| "%Y-%m-%d"                     | 2026-03-05                          |
| "%d-%m-%Y"                       | 05-03-2026                            |
| "%m/%d/%Y"                         | 03/05/2026                              |
| "%d/%m/%Y"                           | 05/03/2026                                |
| "%Y-%m-%d %H:%M:%S"                    | 2026-03-05 14:30:45                         |
| "%B %d, %Y"                              | March 05, 2026                                |
