| Escape | Meaning                     | Example                  | Displays as       |
|--------|-------------------------------|---------------------------|----------------------|
| \n     | New line                      | "a\nb"                    | a  (newline)  b      |
| \t     | Tab (horizontal space)        | "a\tb"                    | a    b               |
| \"     | Literal double quote           | "She said \"hi\""         | She said "hi"        |
| \'     | Literal single quote            | 'It\'s'                   | It's                 |
| \\     | Literal backslash              | "C:\\\\data"               | C:\data              |
| \r     | Carriage return (rare, old-school line reset) | "a\rb" | overwrites start of line |

| Parameter | Purpose                          | Default   | Example                          |
|-----------|-------------------------------------|-----------|-------------------------------------|
| sep       | Separator between multiple values    | " " (space) | print("a","b",sep="-") → a-b     |
| end       | What's added after the print statement | "\n" (newline) | print("hi",end="!") → hi!    |

```python
# string escape sequenece characters:

print("The sum of \"a\":", a, "and b:", b, "is = ", a+b)

print("The sum of \'a\':", a, "and b:", b, "is = ", a+b)

print('The sum of \'a\':', a, 'and b:', b, 'is = ', a+b)

print('The sum of "a":', a, 'and b:', b, 'is = ', a+b)

# \ - to print backslash
print("C:\\Users\\npbha\\OneDrive")

# \n - to print in new line
print("Hello\n Hey")

# \r - carriage return (overwrites the start of the line)
print("Hello\rHey")

# \t - tab space
print("Hello\tHey")
```

# Logical Operators    and, or, not

False and True ---> False
True and True  ---> True
True and False ---> False
False and False ---> False

True or False --> True
True or true ---> True
False or true ---> True
False or false ---> False

not True ---> False
not False ---> True

True xor True ---> False 
True xor False ---> True
False xor True ---> True
False xor False ---> False

```python
print(a==b and a<b)

print(a==b or a<b)

print(a==b | a<b)

print(a != b)
```

# Bitwise Operators  &, |, ~, ^, <<, >>(and, or, not, xor, left-shift, right-shift)

```python
a = 5    # 0101
b = 7  # 0111
print(a&b)
print(a|b)
print(a^b)
print(~a)   # negation uses 2's complement formula ----> -(n+1)
print(a>>1)
print(a<<1)
print(a>>2)
print(a<<2)
```

# F-string Formatting

```python
print(f"The first element in a is {a[0]} ")
print(f"The first element is {a[0]} and second element is {a[1]}")
```

```python
name = input("Enter your Name")
print(f"The customers name was {'name'}")
print(f"The customers name was {name}")
print(f"The customers name was name")
```

| Symbol | Meaning                              | Example                | Output       |
|--------|------------------------------------------|---------------------------|----------------|
| {}     | Placeholder — inserts a variable's value  | f"{name}"                  | Wu             |
| .nf    | Round decimal to n places                 | f"{3.14159:.2f}"           | 3.14           |
| <      | Left-align within a given width           | f"{5:<5}\|"                | 5    \|         |
| >      | Right-align within a given width          | f"{5:>5}\|"                |     5\|         |
| ^      | Center-align within a given width         | f"{5:^5}\|"                |   5  \|         |
| ,      | Thousands separator                        | f"{1234567:,}"             | 1,234,567      |
| b      | Convert to binary                          | f"{10:b}"                  | 1010           |
| o      | Convert to octal                           | f"{10:o}"                  | 12             |

```python
Bill_amount = 86700.678040232
print(f"The final bill is {Bill_amount: .2f}")     # .2f means after decimal 2 places
print(f"The final bill is {Bill_amount:,}") 
print(f"The final bill is { 86700.678040232: .3f}") 
print(f"The binary value of 8 is {8:b}")
print(f"The octal value of 8 is {8:o}")
print(f"The final bill value is {Bill_amount: .1%}")
print(f"The final bill value is {Bill_amount:,.2f}")
```

# Alignment & Padding 

```python
# Alignment & Padding ---> for giving space left and right 

print(f"{'Name': <10} {'score': >8}")
print(f"{'Aisha': <10} {98.7876: >8}")
print(f"{'Aisha': <10} {98.7876: >8.2f}")
print(f"{'Rohan': <10} {67.37876: >8.2f}")

print(f"| {'Matt Murdock': ^25}|")   # center alignment
print(f"| {'Matt Murdock':025}|")  # alignment is by default to left for text data
print(f"| {34000:015}|")    # alignment is by default for a number is right

print(f"| {'Matt Murdock':*>25}|")
print(f"| {'Matt Murdock':&<25}|")
print(f"| {'Matt Murdock':*^25}|")
```

# Precedence and Associativity

An *expression* is anything that evaluates to a value — 2 + 3, x * y, True and x > 5.

### Precedence Table (High → Low)

| Priority | Operator(s) | Description |
|----------|-------------|-------------|
| 1 | () | Parentheses |
| 2 | ** | Exponentiation |
| 3 | +x -x | Unary sign |
| 4 | * / // % | Multiply / Divide |
| 5 | + - | Add / Subtract |
| 6 | << >> | Bitwise shift |
| 7 | & ^ \| | Bitwise AND/XOR/OR |
| 8 | == != > < >= <= | Comparison |
| 9 | not | Logical NOT |
| 10 | and | Logical AND |
| 11 | or | Logical OR |

*Associativity:* most operators → left to right · ** → right to left (only exception)

```python
print(4+6*10-17**4/6)

print(4+6*10-17**(4/6))

print(5**6**2)

print(3**2**3)
```

```python
print(10 and 12)
print(15 or 15)
print(not 15)
```
