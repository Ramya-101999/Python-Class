| Escape | Meaning                     | Example                  | Displays as       |
|--------|-------------------------------|---------------------------|----------------------|
| \n     | New line                      | "a\nb"                    | a  (newline)  b      |
| \t     | Tab (horizontal space)        | "a\tb"                    | a    b               |
| \"     | Literal double quote           | "She said \"hi\""         | She said "hi"        |
| \'     | Literal single quote            | 'It\'s'                   | It's                 |
| \\     | Literal backslash              | "C:\\\\data"               | C:\data              |
| \r     | Carriage return (rare, old-school line reset) | "a\rb" | overwrites start of line |

| Parameter | Purpose                          | Default   | Example                          |
|-----------|-------------------------------------|-----------|-------------------------------------|
| sep       | Separator between multiple values    | " " (space) | print("a","b",sep="-") → a-b     |
| end       | What's added after the print statement | "\n" (newline) | print("hi",end="!") → hi!    |

```python
# string escape sequenece characters:

print("The sum of \"a\":", a, "and b:", b, "is = ", a+b)

print("The sum of \'a\':", a, "and b:", b, "is = ", a+b)

print('The sum of \'a\':', a, 'and b:', b, 'is = ', a+b)

print('The sum of "a":', a, 'and b:', b, 'is = ', a+b)

# \ - to print backslash
print("C:\\Users\\npbha\\OneDrive")

# \n - to print in new line
print("Hello\n Hey")

# \r - carriage return (overwrites the start of the line)
print("Hello\rHey")

# \t - tab space
print("Hello\tHey")
```

```python
# Parameters to use with print function

# step - separator
print("This", "is", "python", "class") 
print("This", "is", "python ", "class", sep = '@')
print("Mango","Banana", "Apple", "Orange", sep ="|")
```

```python
# end  - end of the string
print("Hello Welcome to our Home", end = "...")
print("Hi")
print("Hello")
```

# Operators

# Arithmetic Operators   +, -, *, /, %, //, **

```python
a = 7
b = 10

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)    # floor division ignores the decimal part
print(a%b)   # modulus gives remainder value
print(a**b)  # exponentiation gives power of a value
```

# Assignment Operators =, +=, -=, *=, /=, //=, **=, %=  

```python
a = 7
a+=9
print(a)

a-=10
print(a)

red = 9
red+=10
print(red)
```

# Comparision Operators  ==, !=, >, <, >=, <=

```python
a = 10
b = 5

print(a==b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a!=b)
```

# Logical Operators    and, or, not

False and True ---> False
True and True  ---> True
True and False ---> False
False and False ---> False

True or False --> True
True or true ---> True
False or true ---> True
False or false ---> False

not True ---> False
not False ---> True

```python
print(a==b and a<b)

print(a==b or a<b)

print(a==b | a<b)

print(a != b)
```
