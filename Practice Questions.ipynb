 Q3 (Return multiple values):
 Write get_workout_stats(reps_list) that returns both the total and
 average reps from a list. Unpack both when calling it.

```python
def get_workout_stats(reps_list):
    """
    Calculate total and average reps from a list.
    
    Returns:
        total_reps (int): sum of all reps
        avg_reps (float): average reps
    """
    total_reps = sum(reps_list)
    avg_reps = total_reps / len(reps_list) if reps_list else 0
    return total_reps, avg_reps


reps = [10, 12, 15, 20]
total, average = get_workout_stats(reps)

print("Total reps:", total)    
print("Average reps:", average) 
```

 Q4 (print vs return):
 Write print_bmi(weight, height) that PRINTS the BMI, and return_bmi(weight, height)
 that RETURNS it. Show that only the return version lets you add 1 to
 the result afterward.

```python
def print_bmi(weight, height):
    """
    Prints the BMI directly.
    Formula: BMI = weight / (height^2)
    """
    bmi = weight / (height ** 2)
    print("BMI:", bmi)

def return_bmi(weight, height):
    """
    Returns the BMI value.
    Formula: BMI = weight / (height^2)
    """
    return weight / (height ** 2)


print_bmi(70, 1.75)   

result = return_bmi(70, 1.75)
print("Returned BMI:", result)       
print("Returned BMI + 1:", result+1)  
```

 Q5 (Docstring):
 Write calculate_bmi(weight_kg, height_m) with a Google-style docstring
 (Args, Returns), returning weight_kg / height_m**2. Print its _doc_.

```python
def calculate_bmi(weight_kg, height_m):
    """
    Calculate Body Mass Index (BMI).

    Args:
        weight_kg (float): Weight in kilograms.
        height_m (float): Height in meters.

    Returns:
        float: The BMI value calculated as weight_kg / (height_m ** 2).
    """
    return weight_kg / (height_m ** 2)

print(calculate_bmi.__doc__)
print("BMI:", calculate_bmi(70, 1.75))
```

 Write classify_intensity(heart_rate) returning "Low" (<100), "Moderate"

```python
def classify_intensity(heart_rate):
    """
    Classify workout intensity based on heart rate.
    
    Returns:
        "Low" if heart_rate < 100
        "Moderate" if heart_rate < 140
        "High" otherwise
    """
    if heart_rate < 100:
        return "Low"
    if heart_rate < 140:
        return "Moderate"
    return "High"

print(classify_intensity(85))   
print(classify_intensity(120)) 
print(classify_intensity(160))  
```

 Q10 (**kwargs):
 Write member_profile(**details) that prints whatever keyword details
 are passed in. Call it with name, age, and goal.

```python
def member_profile(**details):
    """
    Print whatever keyword details are passed in.
    
    Args:
        **details: Arbitrary keyword arguments representing member info.
    """
    for key, value in details.items():
        print(f"{key}: {value}")


member_profile(name="Alice", age=25, goal="Build strength")
print("---")
member_profile(name="Bob", age=30, goal="Lose weight")
```

 Given weights = [20, 10, 10], call the function using * to unpack
 the list directly.

```python
def calculate_total_weight(bar, plate1, plate2):
    """
    Calculate the total weight by summing bar and two plates.
    """
    return bar + plate1 + plate2

# Example list of weights
weights = [20, 10, 10]

# Unpack the list directly into the function
total = calculate_total_weight(*weights)

print("Total weight:", total)  
```

 Q15 (Looping through *args and **kwargs inside a function):
 Write show_all(*args, **kwargs) that loops through args printing each
 as "exercise: <value>", and loops through kwargs.items() printing
 each as "<key>: <value>".

```python
def show_all(*args, **kwargs):
    """
    Show all exercises (positional args) and notes (keyword args).
    """
    # Loop through positional arguments
    for exercise in args:
        print(f"exercise: {exercise}")
    
    # Loop through keyword arguments
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_all("Push-ups", "Squats", duration="30 min", mood="Energetic")
print("---")
show_all("Plank", "Burpees", "Lunges", duration="20 min", intensity="High")
```

# H3 (Filtering a list of dicts with two combined conditions):
catalog = [{"title": "Dune 2", "genre": "Sci-Fi", "score": 8.5},
           {"title": "Barbie", "genre": "Comedy", "score": 6.9},
           {"title": "Interstellar", "genre": "Sci-Fi", "score": 8.7},
           {"title": "Anyone But You", "genre": "Comedy", "score": 6.0}]
Build a list of titles that are BOTH genre "Sci-Fi" AND scored 8.0 or above.

```python
catalog = [{"title": "Dune 2", "genre": "Sci-Fi", "score": 8.5},
           {"title": "Barbie", "genre": "Comedy", "score": 6.9},
           {"title": "Interstellar", "genre": "Sci-Fi", "score": 8.7},
           {"title": "Anyone But You", "genre": "Comedy", "score": 6.0}]

titles = []

for movie in catalog:
    if movie["genre"] == "Sci-Fi" and movie["score"] >= 8.0:
        titles.append(movie["title"])

print(titles)
```
