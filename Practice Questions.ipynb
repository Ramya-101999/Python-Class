 21. Is the bestsellers list mutable or immutable? Explain using what
 you just did in questions 18-20 as evidence.

Answer:
The bestsellers list is mutable.

Evidence from what you did in Q18–Q20:

In Q18, you used .remove("Sapiens") to delete an item from the list.

In Q20, you directly replaced "Ikigai" with "Ikigai (Updated Edition)" using index assignment.

Both of these operations changed the original list in place. If lists were immutable (like strings or tuples), you wouldn’t be able to modify, remove, or replace elements directly — you’d have to create a new list instead.

So, the fact that you could remove and update elements proves that Python lists are mutable data structures.

 Q10 (List Methods + Mutability):
 Given lineup = ["Uche", "Lin", "Petra"]:
   a) Append "Sam", remove "Lin", insert "Divya" at position 0. Print lineup.
   b) Set backup = lineup (not a copy!), then append "Omar" to backup.
      Print lineup again — did it change too? Check with ==

```python
lineup = ["Uche", "Lin", "Petra"]
lineup.append("Sam")
lineup.remove("Lin")
lineup.insert(0,"Divya")
print(lineup)


backup = lineup
lineup.append("Omar") 
print(lineup)
print(backup)
print(lineup == backup)
```
