# Slicing the list

slicing = [start:stop:step]

```python
lst = [10, 20, 30, 45.6, True, False,56, 21, 6.6, True]
print("current elements in the list: ", lst)
slice1 = lst[2:5]               # 1st element is start and 2nd element is stop
print("slice 1: ", slice1)
slice2 = lst[5:9]
print("slice 2: ", slice2)
slice3 = lst[:4]
print("slice 3: ", slice3)
slice4 = lst[3:]
print("slice 4: ", slice4)
slice5 = lst[:]
print("slice 5: ", slice5)
```

```python
print("original list: ", lst)    # By default the stepcount in list is 1
slice6 = lst[::2]
print("slice 6 (every second element): ", slice6)
```

# Negative Indexing And Slicing

```python
# Negative Indexing and slicing
lst = [10, 20, 30, 45.6, True, False,56, 21, 6.6, True]
print("original list: ", lst)
slice7 = lst[-5:-2]
print("slice 7 (from index -5 to -2): ", slice7)
slice8 = lst[-2:-5 : -1]        # if 1st parameter is greater than 2nd one then we need to give 3rd parameter
print("slice 8 (from index -2 to -5): ", slice8)
```

## Python List Methods and Their Return Types

| List Method | Return Type | Description |
|-------------|-------------|-------------|
| append(x) | None | Adds an element to the end of the list. |
| extend(iterable) | None | Adds all elements from an iterable to the list. |
| insert(index, x) | None | Inserts an element at the specified index. |
| remove(x) | None | Removes the first occurrence of the specified element. |
| pop([index]) | Any | Removes and returns the element at the given index (last element by default). |
| clear() | None | Removes all elements from the list. |
| index(x[, start[, end]]) | int | Returns the index of the first occurrence of an element. |
| count(x) | int | Returns the number of occurrences of an element. |
| sort(key=None, reverse=False) | None | Sorts the list in place. |
| reverse() | None | Reverses the list in place. |
| copy() | list | Returns a shallow copy of the list. |

### Return Type Summary

| Return Type | Methods |
|-------------|---------|
| None | append(), extend(), insert(), remove(), clear(), sort(), reverse() |
| int | index(), count() |
| list | copy() |
| Any | pop() |

# .deepcopy() method

 When inner list is present copy method fails so we prefer deepcopy
 Where affecting one list should not affect the other but it is failing to show that

 copy method always creates a shallow copy
 When having an inner list we should always go for creating a deep copy

```python
import copy
i =  [1,2,4,5,[8,9,10]]
j = copy.deepcopy(i)

i[4][0] = 100
print(i)
print(j)
```
