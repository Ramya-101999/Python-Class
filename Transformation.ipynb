# Transformation 

```python
# Transformation - Data transformation using .map() - for dictionary substitution,
# .apply() for custom functions on rows and columns

# type conversion  - astype() and handling errors with pd.to_numeric(errors='coerce')
```

```python
import os 
os.chdir("Data Sets")
```

```python
import pandas as pd

df = pd.read_csv("all_athlete_games.csv", low_memory=False)
sample = df[df["Medal"].notna()][["Name","Gender","Age","NOC","Sport","Medal","Year"]].head(50).reset_index(drop=True)
print(sample)
```

```python
# mapping, .map() for dictionary substitution

#syntax - series.map(dict/function)

#where mapping is a dict or a funtion

points_map = {"Gold":3,"Silver":2,"Bronze":1}
print(sample["Medal"].map(points_map))
print(sample["Medal"])

gender_map = {"Male":"M"}
print(sample["Gender"].map(gender_map))
print(sample["Gender"])
```

```python
# mapping with a function instead of a dict

print(sample["Medal"].map(len))
```

```python
# veteran if age>=30 else young 

print(sample["Age"].map(lambda a:"Veteran" if a>=30 else "Young"))
```

```python
# .apply()

# for applying custom on rows/columns

# syntax - series.apply(function)

# df.apply(function,axis=0 or 1)

# .apply on a single column(series)

print(sample["Age"].apply(lambda a:round(a*10)))
```

```python
# row wise apply on a dataframe

def summarize(row):
    return f"{row["Name"]}({row["NOC"]}) won {row["Medal"]} in {row["Sport"]}"

print(sample.apply(summarize,axis=1))

print()

# column-wise apply (default method)

print(sample[["Age"]].apply(lambda col:col.max()-col.min()))
```

| Call                          | What gets passed to your function              |
|------------------------------------|--------------------------------------------------------|
| series.apply(func)                     | Each individual VALUE, one at a time                       |
| df.apply(func, axis=0)  (default)         | Each entire COLUMN, one at a time                             |
| df.apply(func, axis=1)                       | Each entire ROW, one at a time — access fields by row["col"]     |

# Type Conversion

```python
# Type conversion - astype()

# and pd.to_numeric(errors='coerce')

# astype() - straight forward conversion when data is already clean

print(sample["Age"].astype(int))

# astype is intolerant of bad data, does not convert or fails to run the moment it finds a null value
```

```python
print(sample["Year"] if False else df["Year"].astype(str).head(3))
```

```python
# pd.to_numeric(errors='coerce') - the bad values become NaN instead of crashing 

sample["Age"] = pd.to_numeric(sample["Age"],errors="coerce")
print(sample["Age"])
```

```python
# errors = 'raise' works the same way as astype conversion with null values in the column
sample["Age"] = pd.to_numeric(sample["Age"],errors="raise")
print(sample["Age"])
```
