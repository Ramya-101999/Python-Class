```python
# Transformation - Data transformation using .map() - for dictionary substitution,
# .apply() for custom functions on rows and columns

# type conversion  - astype() and handling errors with pd.to_numeric(errors='coerce')
```

```python
import os 
os.chdir("Data Sets")
```

```python
import pandas as pd

df = pd.read_csv("all_athlete_games.csv", low_memory=False)
sample = df[(df["Year"]==2020) & (df["Medal"].notna())][["Name","Gender","Age","NOC","Sport","Medal"]].head(8).reset_index(drop=True)
print(sample)
```

```python
# mapping, .map() for dictionary substitution

#syntax - series.map(dict/function)

#where mapping is a dict or a funtion

points_map = {"Gold":3,"Silver":2,"Bronze":1}
print(sample["Medal"].map(points_map))
print(sample["Medal"])

gender_map = {"Male":"M"}
print(sample["Gender"].map(gender_map))
print(sample["Gender"])
```

```python
# mapping with a function instead of a dict

print(sample["Medal"].map(len))
```

```python
# veteran if age>=30 else young 

print(sample["Age"].map(lambda a:"Veteran" if a>=30 else "Young"))
```

```python
# .apply()

# for applying custom on rows/columns

# syntax - series.apply(function)

# df.apply(function,axis=0 or 1)

# .apply on a single column(series)

print(sample["Age"].apply(lambda a:round(a*10)))
```
