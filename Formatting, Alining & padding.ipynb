```python
# BMI Calculator 

emp_name = str(input("Enter the employee name: "))
weight = float(input("Enter weight in kilograms: "))
height = float(input("Enter height in meters: "))

bmi = weight / (height ** 2)

print("Employee Name:", emp_name)
print("Weight:", weight, "kg")
print("Height:", height, "m")
print("BMI:", bmi)

print("BMI + 2:", bmi + 2)
print("BMI - 1:", bmi - 1)
print("BMI doubled:", bmi * 2)
print("Half of BMI:", bmi / 2)
```

```python
# age calculations

emp_name = str(input("Enter the employee name: "))
dob = str(input("Enter date of birth (YYYY-MM-DD): "))
age = int(input("Enter your age: "))

print("Employee Name:", emp_name)
print("Date of Birth:", dob)
print("Age:", age)

print("Age + 5:", age + 5)
print("Age - 2:", age - 2)
print("Age doubled:", age * 2)
print("Half of Age:", age / 2)
```

```python
# Rectangle Area

length = float(input("Enter rectangle length: "))
width = float(input("Enter rectangle width: "))

area = length * width

print("Length:", length)
print("Width:", width)
print("Area:", area)

print("Area + 10:", area + 10)
print("Area - 5:", area - 5)
print("Area doubled:", area * 2)
print("Half of Area:", area / 2)
```

```python
# Shopping Bill calculations

item_name = str(input("Enter item name: "))
price = float(input("Enter item price: "))
quantity = int(input("Enter quantity: "))

bill = price * quantity

print("Item:", item_name)
print("Price:", price)
print("Quantity:", quantity)
print("Total Bill:", bill)

print("Bill + 50:", bill + 50)
print("Bill - 20:", bill - 20)
print("Bill doubled:", bill * 2)
print("Half of Bill:", bill / 2)
```

```python
# Student enrollment

student_name = str(input("Enter student name: "))
course = str(input("Enter course name: "))
fees = float(input("Enter course fees: "))

print("Student Name:", student_name)
print("Course:", course)
print("Fees:", fees)

print("Fees + 1000:", fees + 1000)
print("Fees - 500:", fees - 500)
print("Fees doubled:", fees * 2)
print("Half of Fees:", fees / 2)
```

```python
# employee enrollment

emp_name = str(input("Enter employee name: "))
department = str(input("Enter department: "))
salary = float(input("Enter salary: "))

print("Employee Name:", emp_name)
print("Department:", department)
print("Salary:", salary)

print("Salary + 2000:", salary + 2000)
print("Salary - 1000:", salary - 1000)
print("Salary doubled:", salary * 2)
print("Half of Salary:", salary / 2)
```

```python
# Movie ticket booking

movie_name = str(input("Enter movie name: "))
ticket_price = float(input("Enter ticket price: "))
tickets = int(input("Enter number of tickets: "))

total = ticket_price * tickets

print("Movie:", movie_name)
print("Ticket Price:", ticket_price)
print("Tickets:", tickets)
print("Total Cost:", total)

print("Cost + 100:", total + 100)
print("Cost - 50:", total - 50)
print("Cost doubled:", total * 2)
print("Half of Cost:", total / 2)
```

```python
# Pizza Orders

pizza_type = str(input("Enter pizza type: "))
price = float(input("Enter pizza price: "))
quantity = int(input("Enter number of pizzas: "))

bill = price * quantity

print("Pizza Type:", pizza_type)
print("Price:", price)
print("Quantity:", quantity)
print("Total Bill:", bill)

print("Bill + 100:", bill + 100)
print("Bill - 50:", bill - 50)
print("Bill doubled:", bill * 2)
print("Half of Bill:", bill / 2)
```

```python
# Fun activity bills

activity = str(input("Enter activity name: "))
cost = float(input("Enter cost per person: "))
people = int(input("Enter number of people: "))

total = cost * people

print("Activity:", activity)
print("Cost per person:", cost)
print("Number of People:", people)
print("Total Cost:", total)

print("Cost + 200:", total + 200)
print("Cost - 100:", total - 100)
print("Cost doubled:", total * 2)
print("Half of Cost:", total / 2)
```
