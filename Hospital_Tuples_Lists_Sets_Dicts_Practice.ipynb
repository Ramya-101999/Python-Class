# PART 3 — SETS

```python
blood_entries = ["A+", "B+", "O+", "A+", "AB+", "B+", "O-"]
morning_shift = {"Dr. Sharma", "Dr. Patel", "Dr. Rao", "Dr. Nair"}
evening_shift = {"Dr. Rao", "Dr. Nair", "Dr. Mehta"}
```

```python
# 3a. Convert blood_entries to a set and print the unique blood groups
blood_entries = ["A+", "B+", "O+", "A+", "AB+", "B+", "O-"]
unique_blood_groups= set(blood_entries)
print(unique_blood_groups)
```

```python
# 3b. How many duplicate entries were there? (compare the two lengths)
duplicates = len(blood_entries) - len(unique_blood_groups)
print(duplicates)
```

```python
# 3c. Add "Dr. Khan" to morning_shift
morning_shift.add("Dr. khan")
print(morning_shift)
```

```python
# 3d. Remove "Dr. Patel" using the method that won't crash if missing
morning_shift.discard("Dr. Patel")
print(morning_shift) 
```

```python
# 3e. Print doctors who worked BOTH shifts (intersection)
print(morning_shift.intersection(evening_shift))
```

```python
# 3f. Print doctors who worked EITHER shift (union)
print(morning_shift.union(evening_shift))
```

```python
# 3g. Print doctors who worked ONLY the morning shift (difference)
print(morning_shift.difference(evening_shift))
```

```python
# 3h. Print doctors who worked exactly one shift, not both (symmetric_difference)
print(morning_shift.symmetric_difference(evening_shift))
```
