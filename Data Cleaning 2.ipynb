```python
import pandas as pd
pd.set_option('display.max_columns',None)
df = pd.read_csv("all_athlete_games.csv", low_memory=False)
sample = df[df["Year"]==2020] [["Entry ID","Name","Team","Sport","Event","Medal"]].head(12).reset_index(drop=True)
print(sample[["Name","Team","Sport","Event"]])
```

# Duplicates

```python
# Duplicates

# drop_duplicates(subset,keep)

# check with duplicated() first

# syntax - df.duplicated(subset=None,keep='first')

print(sample.duplicated())
print()
print(sample.duplicated(subset="Name"))
```

```python
# Quick counting of duplicates

print(sample.duplicated(subset="Name").sum())

# Checking multiple columns at once

print(sample[sample.duplicated(subset=["Name","Sport"],keep="first")]
      [["Name","Sport","Event"]])

# keep = False in duplicated is to view every row in a duplicate group(not for dropping)
```

| keep=          | Behavior                                              |
|---------------------|--------------------------------------------------------------|
| 'first' (default)       | Keep the first occurrence, drop later repeats                    |
| 'last'                     | Keep the last occurrence, drop earlier repeats                     |
| False                        | Drop EVERY row that has any duplicate at all                         |

```python
# removing with drop_duplicates()

# drop_duplicates(subset,keep)

print(sample.drop_duplicates(subset="Name"))
```

```python
print(sample.drop_duplicates(subset="Name",keep='last'))
```

```python
print(sample.drop_duplicates(subset="Name",keep='first'))
```

# String Operations

```python
# String operations : The .str accessor

# .str unlocks string methods on an entire column at once - the same methods you already know, vectorized across every row

# Core methods

# .str.lower()

print(sample["Name"].str.lower())

# without .str calling .lower() directly on a series would throw error.

# .str applies a string method to every row
```

```python
# .str.contains()

# syntax - series.str.contains(patterns,case=True,na=False,regex=True)

print(sample[sample["Event"].str.contains("Team")] [["Name","Event"]])
```

```python
print(sample[sample["Event"].str.contains("team",case=False)] [["Name","Event"]])

# case = False makes the search case insensitive
```

```python
# .str.len(), .str.strip(), .str.split()

a = "hi there"
print(a.split())

print(sample["Name"].str.len())
print()
print(sample["Name"])
print()
print(sample["Name"].str.split().str[0])
```

```python
messy=pd.Series(["  Ananya  ","Rahul  ","  Priya"])

print(messy.str.strip())
```

```python
# str.replace()

print(sample["Event"].str.replace("Men's","mens"))
print(sample["Event"].str.replace("Men's","mens",regex=False))

# regex = False treats the search text literally not as a pattern
```

```python
# .str.title()

messy_case = pd.Series(["MEN'S SINGLES", "women's doubles"])
print(messy_case.str.title())

# .title() capitalizes after any non-letter character not just spaces

# .str.count()

print(sample["Event"].str.count("'"))

# .str[i] - indexing into each string like a list

print(sample["Team"].str[0])
```

```python
# .str.slice()

print(sample["Sport"].str.slice(0,4))

# .str.cat() - joining string columns together

print(sample["Name"].str.cat(sample["Team"],sep=" - ").str.cat(sample["Event"],sep=" - "))
```

```python
# Chaining multiple .str methods

print(sample["Sport"].str.lower().str.replace(" ","_"))
print(sample)
```

```python
# Regex [Regular expression]

# Basic patterns for extracting data from text
# Regex lets you match patterns in text, not just exact substrings - for pulling structured data out of messy text fields 
```

| Pattern | Meaning                                    |
|-------------|----------------------------------------------|
| \d              | Any single digit (0-9)                            |
| \w                | Any "word" character (letters, digits, underscore)   |
| \s                  | Any whitespace character                                |
| ^                     | Anchors to the START of the string                        |
| $                       | Anchors to the END of the string                             |
| +                         | One or more of the preceding pattern                            |
| {n}                          | Exactly n repeats of the preceding pattern                          |
| \|                              | OR — matches any one of several alternatives                          |

```python
# \d - matching digits

events_with_weight = df[df["Event"].str.contains(r"\d+ ?kg", regex=True, na=False)][["Event"]].head(5)
print(events_with_weight)
```

```python
# str.extract() - pulling the matched value but

# syntax - series.str.extract(pattern)

# -wrap the part to capture in ()

weight_extract = events_with_weight["Event"].str.extract(r"(\d+)\s?kg")
print(weight_extract)
```

 | Quantifier | Meaning                                    | Example                     |
|----------------|-----------------------------------------------|----------------------------------|
| ?                  | Zero or ONE of the preceding character            | " ?" → space is optional            |
| +                    | One or MORE of the preceding character              | "\d+" → one or more digits            |
| *                      | Zero or MORE of the preceding character                | " *" → any number of spaces, including none |
| {n}                      | EXACTLY n of the preceding character                     | "\d{3}" → exactly 3 digits                  |
weight_extract = events_with_weight["Event"].str.extract(r"(\d+)\s?kg")
print(weight_extract)

```python
# ^ and $ - start and end anchors

print(sample[sample["Name"].str.contains(r"^A", regex=True)][["Name"]])
print(sample[sample["Event"].str.contains(r"Team$", regex=True)][["Event"]])
```

```python
# Combining ^ and $ - the exact whole string match

test = pd.Series(["Team", "Men Team", "Teamwork"])
print(test.str.contains(r"^Team$",regex=True))

# na can be only None, pd.NA, np.nan, True, Or false
```

```python
# na

s_obj = pd.Series(["Team Event","Solo Event",None,"Mixed Team"],dtype="object")
print(s_obj.str.contains("Team"))

s_obj = pd.Series(["Team Event","Solo Event",None,"Mixed Team"],dtype="object")
print(s_obj.str.contains("Team",na=False))
```

| Parameter    | What it controls                                            |
|------------------|----------------------------------------------------------------------|
| regex=True (default)  | Pattern is interpreted as a REGEX (special characters like \s, \d work) |
| regex=False              | Pattern is treated as a LITERAL string (no special regex meaning)          |
| na=False                   | Missing values produce False in the result, instead of NaN/None               |
| na left unset                 | Missing values produce NaN/None in the result (can break masking)                |

```python
# \s - whitespace
print(df[df["Event"].str.contains(r"Team\s", regex=True, na=False)][["Event"]].head(3))
```

```python
# {n} - exact repeat count
print(df[df["Event"].str.contains(r"\d{3}", regex=True, na=False)][["Event"]].drop_duplicates().head(5))
```

```python
import pandas as pd

df = pd.read_csv("all_athlete_games.csv", low_memory=False)
winter = df[(df["Season"]=="Winter") & (df["Year"]==2014)][["Entry ID","Name","Gender","Team","Sport","Event","Medal"]].head(15).reset_index(drop=True)
print(winter)
```

```python
# Problem 1 -

# How many rows in 'winter dataframe' are duplicated based on "Name" column

print(winter.duplicated(subset="Name").sum())
```

```python
# Problem 2 - 

# drop duplicate names, keeping only the first event listed for each athlete

winter.drop_duplicates(subset="Name")[["Name","Event"]]
```

```python
# Problem 3 -

# (.str + casing)

# Using .str.lower(), check which row's Name contains "boltia" - this should catch Agnese's rows regardless of her inconsistent capitalization

print(winter[winter["Name"].str.lower().str.contains("boltia")])
```

```python
# Problem 4 -

# (.str.contains()):

# Filter to all rows where "Sport" contains "Skiing" (any skiing discipline)


print(winter[winter["Sport"].str.contains("Skiing")][["Name","Sport"]])
```

```python
# Problem 5 (Regex — decimal-aware extraction): The Biathlon events contain
#            distances like "10 kilometres" and "12.5 kilometres". Extract
#            just the number (including decimals) from each Biathlon event.
#            Hint: (\d+(?:\.\d+)?) matches whole numbers AND decimals.

biathlon = winter[winter["Sport"]=="Biathlon"][["Event"]]
print(biathlon["Event"].str.extract(r"(\d+(?:\.\d+)?) kilometers"))
print(biathlon)

# (\d+(?:\.\d+)?)

# (...........) - outer capture group to extract the pattern
# \d+ - one or more digits - matches the whole number part

# (?:......) - a non-capture group - groups pieces together without extracting them separately 

#  \. - a decimal point

# ? -  makes the whole (?:\.\d+) group optional - zero or one occurence
```

```python
# Problem 6 (Regex — relay pattern): Filter to events matching a relay
#            pattern like "2 x 6" (a number, space, x, space, number).
print(winter[winter["Event"].str.contains(r"\d+\s+x\s+\d+")][["Event"]])
```

```python
# Problem 7 (Basic pandas): How many DISTINCT sports appear in winter?

print(winter["Sport"].nunique())
```

```python
# Problem 8 (Boolean masking): Filter to only rows where a Medal was won.

print(winter[winter["Medal"].notna()] [["Name","Team","Event","Medal"]])
```
