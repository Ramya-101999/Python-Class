# Statistics

There are 4 types of statistics. They are:

        Statistics                     Python tools used

1. Descriptive statistics         -         Pandas, Numpy
2. predictive statistics          -        scipy.stats,statsmodels
3. prescriptive statistics        -        scikit-learn, statsmodels             
4. Inferential statistics          -       scipy.optimize + predictivemodels, specialized tools

| Type                     | What it does                                                | Example question                          |
|-------------------------------|----------------------------------------------------------------------|--------------------------------------------------|
| Descriptive Statistics             | Summarizes and describes data you ALREADY HAVE                          | "What's the average age of our athletes?"            |
| Inferential Statistics                | Uses a SAMPLE to draw conclusions about a larger POPULATION                | "Based on 500 surveyed athletes, what's likely true for all Olympic athletes?" |
| Predictive Statistics                    | Uses past data to forecast FUTURE outcomes                                  | "Based on training data, will this athlete medal next year?"    |
| Prescriptive Statistics                     | Recommends an ACTION based on data and predictions                           | "Given the trends, which athletes should get more funding?"        |

# Descriptive Statistics

It is a toolkit of a handful of single numbers that each summarize something important about a whole pile of data, so you can understand it at a glance without scrolling through thousands of rows.

Two big questions statistics helps answer about any group of numbers- 

1. Where is the center of this data? (This is called central tendency)

2. How spread out or clustered together are the values? (This is called dispersion)

Everything going forward is just tools for answering these two questions - plus a systematic way to spot the unusual values that don't fit either picture.
