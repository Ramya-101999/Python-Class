# Descriptive Statistics

It is a toolkit of a handful of single numbers that each summarize something important about a whole pile of data, so you can understand it at a glance without scrolling through thousands of rows.

Two big questions statistics helps answer about any group of numbers- 

1. Where is the center of this data? (This is called central tendency)

2. How spread out or clustered together are the values? (This is called dispersion)

Everything going forward is just tools for answering these two questions - plus a systematic way to spot the unusual values that don't fit either picture.

```python
import os
os.chdir("Data Sets")
```

```python
import pandas as pd

df = pd.read_csv("all_athlete_games.csv", low_memory=False)
sample = df[(df["Year"]==2020) & (df["Medal"].notna())][["Name","Age","Sport"]].dropna(subset=["Age"]).reset_index(drop=True)
ages = sample["Age"]
print(ages.describe())
```

# Part 1 - Central Tendency

```python
# Central Tendency

# Mean ('The average')

print("Mean:",ages.mean())

# Every single athlete's age contributed to this number - the mean is influenced by every value including the extreme ones


# Median ('The middle value')

print("Median:",ages.median())
```

```python
normal_group = pd.Series([25, 27, 28, 26, 29])
print("Mean:", normal_group.mean())
print("Median:", normal_group.median())

# mean and median are close together because of no extreme values
```

```python
with_outlier = pd.Series([25, 27, 28, 26, 90])
print("Mean:", with_outlier.mean())
print("Median:", with_outlier.median())

# one unusual value(90) dragged the mean from 27.0  to 39.2

# The median did not move and this is called "robustness to outliers".

# A statistic is robust if a few wierd extreme values cannot easily distort it.

# Median is robust, mean is not
```

```python
# Dispersion -  How to spread out this data?

# Knowing the center is not the whole pictures.

# Two datasets can share the exact same mean but look completely different.

tight_group = pd.Series([25, 26, 27, 28, 29])
spread_group = pd.Series([5, 15, 27, 40, 48])
print(tight_group.mean(), spread_group.mean())
```

```python
# Range - the simplest measure

print(ages.max()-ages.min())
```

```python
# Variance - average distance from the mean

# for every value measure it's distance from the mean, square that distance (so that negatives out) then average all the squared distances

print(ages.var())

#squaring puts the ages in squared years-which we cannot use for anything so we use std or standard deviation

#Standard deviation - the square root of variance

print(ages.std())
```

```python
tight_group = pd.Series([25, 26, 27, 28, 29])
print(tight_group.mean())
print(tight_group.var())
print(tight_group.std())

spread_group = pd.Series([5, 15, 27, 40, 48])
print(spread_group.mean())
print(spread_group.var())
print(spread_group.std())

# small standard deviation = tightly clustered values

# large standard deviation = scattered, spread out values
```

```python
# Outlier Logic - IQR (Inter Quartile Range) (Q3-Q1) and Whiskers (1.5 x QR)

# Quartiles - Dividing data into 4 equal chunks

# Just like the median splits sorted data in half, quartiles split it into four equal quarters.

# Q1 - the value at the 25% mark

# Q2 - this is just the median (the 50% mark)

# Q3 - the value at the 75% mark

Q1 = ages.quantile(0.25)
Q2 = ages.quantile(0.5)
Q3 = ages.quantile(0.75)

print("Q1:",Q1)
print("Q2:",Q2)
print("Q3:",Q3)
```

```python
# IQR - spread of just the normal middle

IQR = Q3-Q1
print(IQR)

# IQR beats the range for outlier direction:

# Range is defined by the extremes so it cant help find the outliers.

# IQR ignores the outliers that is the extreme top and bottom 255 measuring only the normal middle - a much more stable reference point
```

```python
# Whiskers - the line between "normal" and "outlier"

# Anything more than 1.5 times the IQR beyond Q1 or Q3 counts as an outlier. This 1.5 multiplier is a long standing statistical convention by John Tukey not deliverd from any specific data - just a standard value

lower_whisker = Q1-1.5*IQR
upper_whisker = Q3+1.5*IQR

print("Lower bound:",lower_whisker)
print("Upper bound:",upper_whisker)

# Any athlete younger than 15 or older than 39 is flagged as a statistical outlier. Everyone between 15 and 39 is normal.
```

```python
skateboard = sample[sample["Sport"]=="Skateboarding"]
print(skateboard[["Name","Age"]].sort_values("Age"))
```

```python
print("Mean:",skateboard["Age"].mean())
print("Median:",skateboard["Age"].median())
```

```python
# How spread out is the data

gymnastics = sample[sample["Sport"]=="Artistic Gymnastics"]["Age"]
print("Skateboarding std:",skateboard["Age"].std())
print("Artistic Gymnastics std:",gymnastics.std())
```

```python
# Finding the outliers

outliers = sample[(ages<lower_whisker)|(ages>upper_whisker)]
print(outliers[["Name","Age","Sport"]].sort_values("Age"))
print("Total outliers found:",len(outliers))

# 45 outliers out of 2449 athletes(~1.8%)-5 low end(very young skateboarders/divers),40 high end(overwhelmingly Equestrian, a sport where age doesn't limit performance the way it does anywhere else).
```
