## Section 9 — split() with input()

### Question 9

```python
# 1. Take patient name and ward as input in one line (space separated)
#    Print: "[name] is in [ward] ward"
name, ward = input("Enter the patient name and ward").split(" ")
print(f"{name} is in {ward} ward")
```

```python
# 2. Take patient name, ward, and charge as input in one line
#    Convert charge to integer
#    Print: "[name] is in [ward]. Daily charge: Rs.[charge]"
name, ward, charge = input("Enter the name, ward and charge").split(" ")
charge = int(charge)
print(f"{name} is in {ward} ward and Daily charge is: Rs. 4500")
```

```python
# 3. Take patient id, name, ward, charge as comma separated input
#    Split by comma
#    Print each value on a separate line

id, name, ward, charge = input("Enter the id, name, ward, charge: ").split(",")

print(f"{id}: id")
print(f"{name}: name")
print(f"{ward}: ward")
print(f"{charge}: charge")
```
