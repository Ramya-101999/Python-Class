```python
# read line by line

# every line read from a file includes \n - print() adds its own newline on top causing double break
# so we use .strip() which removes the leftover \n before printing.

f = open("transactions.txt","r")
for line in f:
    print(line.strip())
f.close()
print(os.getcwd())
```

# append mode

```python
f = open("transactions.txt","a")
f.write("Deposit:750\n")
f.close()
```

```python
# context manager - "with" keyword
# To manage the closing of the file easily 
# Here the moment with block ends, automatically python closes the file no .close() needed

with open("transactions.txt","r") as f:
    data = f.read()

print(data)
```

```python
# readline(), readlines()
# you can use readline in multiple calls as it can remember where it stopped last.

with open("transactions.txt","r") as f:
    line1 = f.readline()
    line2 = f.readline()
    line3 = f.readline(4) # to mention how many characters to read
    line4 = f.readline()

print(line1)
print(line2)
print(line3)
print(line4)
```

```python
with open("transactions.txt","r") as f:
    all_lines = f.readlines()

print(all_lines)
print(all_lines[0])
```

```python
# r -  read and write, without erasing the existing content 

with open("transactions.txt","w") as f:
    f.write("ABCDEFGH\n")


with open("transactions.txt","r+") as f:
    content=f.read()
    print(content)
    print(repr(content))
    f.write("XYZ")


with open("transactions.txt","r+") as f:
    content=f.read()
    print(content)
    print(repr(content))  # repr means give the data in its exact form along with any unprintable characters
    f.seek(0)  # to send the cursor to the begining of the file
    f.write("XYZ") # to check what is the cursor position currently
```

```python
# w+ - write and read, but truncates first

with open("transactions.txt","w+") as f:
    f.write("Fresh content\n")
    f.seek(0)
    print(repr(f.read()))
```

```python
# a+ - append and read

with open("transactions.txt","a+") as f:
    f.write("Extra line\n")
    f.seek(0)
    print(repr(f.read()))
```

# Error Handling - try and except

```python
# try
#   risky_code
# except SpecificError:
#   handle_it
# finally
#   always_runs
```

```python
# basic try and except

balance = 1000
withdrawal = 1500

final_balance = balance - withdrawal
print(final_balance)

try:
    if withdrawal > balance:
        raise ValueError("Insufficient funds")
    
except ValueError as e:
    print("Transaction Failed:", e)
```

```python
# Don't allow the user to enter amount > 600

try:
    amount = int(input("Enter the amount"))
    if amount > 600:
        raise ValueError("Your amount is greater than 600")
except ValueError as e:
    print("Invalid amount:", e)
```

```python
# FileNotFoundError

try:
    with open("account_history.txt","r") as f:
        print(f.read())
except FileNotFoundError:
    print("No account history found - have to start fresh")
```

```python
# Multiple except blocks

def process_transaction(amount_str,balance):
    try:
        amount = int(amount_str)
        return balance - amount
    except ValueError:
        return "Error: amount must be a number"
    except TypeError:
        return "Error: Invalid balance type"

print(process_transaction("200",1000))
print(process_transaction("abc",1000))
print(process_transaction("300","abc"))
```

```python
# finally - runs no matter what

try:
    amount = int("zyx")
except ValueError:
    print("Invalid Amount")
finally:
    print("Transaction log closed")
```

| Error Type            | When it happens                                       | Example that triggers it        |
|----------------------------|--------------------------------------------------------------|----------------------------------------|
| ValueError                     | Right type, but an invalid VALUE (e.g. text that isn't a number) | int("abc")                                |
| TypeError                        | Operation used on an incompatible TYPE                            | "5" + 5                                     |
| ZeroDivisionError                  | Dividing any number by zero                                          | 10 / 0                                        |
| IndexError                           | Accessing a list/tuple position that doesn't exist                     | [1,2,3][10]                                     |
| KeyError                               | Accessing a dict key that doesn't exist                                  | {"a":1}["b"]                                      |
| FileNotFoundError                        | Trying to open() a file in "r" mode that doesn't exist                      | open("nope.txt")                                    |
| PermissionError                            | OS denies access — read-only file, locked file, protected folder             | open("locked.txt", "w")                               |
| FileExistsError                              | Using mode "x" on a file that already exists                                    | open("existing.txt", "x")                               |
| AttributeError                                 | Calling a method/attribute that doesn't exist on that object                      | (5).append(1)                                             |
| NameError                                        | Using a variable that was never defined                                             | print(undefined_var)                                        |
| ModuleNotFoundError                                | Trying to import a module that isn't installed/doesn't exist                          | import nonexistent_module                                     |
| IndentationError                                     | Inconsistent or missing indentation                                                     | Mismatched spaces in an if block                                 |
| SyntaxError                                            | Code that breaks Python's grammar rules entirely                                          | if x = 5: (using = instead of ==)                                   |

| Catchable with try/except? | Error types                                                    |
|----------------------------------|--------------------------------------------------------------------|
| Yes — runtime errors                 | ValueError, TypeError, ZeroDivisionError, IndexError, KeyError, FileNotFoundError, PermissionError, FileExistsError, AttributeError, NameError, ModuleNotFoundError |
| No — happens before execution starts    | SyntaxError, IndentationError                                                    |

```python
# catching multiple error types in one except

def safe_convert(value):
    try:
        return 100/int(value)
    except (ValueError, ZeroDivisionError) as e:
        return f"Failed: {e}"

print(safe_convert("5"))
print(safe_convert("abc"))
print(safe_convert("0"))
```

```python
# Only except - catches everything

try:
    result = 10/0
except:
    print("Something is Wrong")
```

```python
# except Exception

try:
    x = [1,2,3]
    print(x[10])
except Exception as e:
    print(f"{type(e).__name__}:{e}")

# type(e).__name__ tells us the exact class that has occured
# useful for logging or debugging when we don't know which eroor might occur
```

```python
# when using multiple types of exception maintain the order - order matters
# specific exceptions should come before the general exception

try:
    value = int("abc")
except ValueError:
    print("Specific: Invalid literal")
except Exception:
    print("General: Something else went wrong")
```
