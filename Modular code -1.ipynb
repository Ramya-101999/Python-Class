# Functions

Custom functions can be defined for repeating tasks and the moment you call, the function will perform it's task.

```python
# SYNTAX:

# def function_name(parameters):
#    task_to_do
#    return value

# syntax to call function:

# function_name(arguments)
```

```python
def add(a,b):
    print(a+b)
add(5,6)
```

```python
# Function without parameter

def greet():
    print("Hello user")
greet()
```

```python
# Function with a parameter

def greetcustom(name):
    print("Hello",name,"Welcome to our store")
greetcustom("Sam")
```

```python
# Function with positional arguments

def total_bill(price,quantity,customer_name):
    total = price*quantity
    print(f"price:{price},Quantity:{quantity},Total:{total},purchased by {customer_name}")

total_bill(5000,4,"Sam") # positional arguments

total_bill(quantity=20,price=67000,customer_name="sam") # keyword arguments

total_bill(6700,5,customer_name="Sam") # positional and keyword arguments

total_bill(6700,customer_name="Sam",quantity=5) # positional and keyword arguments

# When mixing the two types remember all positional arguments must come before the keyword arguments
```

```python
# default parameter

def total_bill(price,quantity,customer_name="Unknown"):
    total = price*quantity
    print(f"price:{price},Quantity:{quantity},Total:{total},purchased by {customer_name}")

total_bill(6700,5)

total_bill(6700,5,"Sam")
```

```python
#  "/" restricts all arguments to be only positional arguments, 
# any parameter mentioned before becomes strictly only positional arguments

def total_bill(price,quantity,customer_name,/,city):
    total = price*quantity
    print(f"price:{price},Quantity:{quantity},Total:{total},purchased by {customer_name},in the city:{city}")

total_bill(5000,5,"Sam",city="Bangalore")
```

```python
#  "*" any parameter after * should strictly be keyword arguments only
def total_bill(price,*,quantity,customer_name,city="Mumbai"):
    total = price*quantity
    print(f"price:{price},Quantity:{quantity},Total:{total},purchased by {customer_name},in the city:{city}")

total_bill(5000,quantity=5,customer_name="Sam",city="Bangalore")
```

```python
def total_bill(price,/,quantity,*,customer_name,city="Mumbai"):
    total = price*quantity
    print(f"price:{price},Quantity:{quantity},Total:{total},purchased by {customer_name},in the city:{city}")

total_bill(5000,quantity=5,customer_name="Sam",city="Bangalore")

total_bill(5000,5,customer_name="Sam",city="Bangalore")

total_bill(5000,5,customer_name="Sam")
```

| Rule                                                  | Applies to                    |
|------------------------------------------------------------|------------------------------------|
| All positional args must come before keyword args (at call time) | Every function call                    |
| No required param can follow a default param (in definition)      | Positional/positional-or-keyword params ONLY |
| Keyword-only params (after *) can freely mix required + default in any order | Keyword-only zone                        |
| Positional-only params (before /) can NEVER be passed by keyword    | Positional-only zone                       |

# Single return and multiple returns

```python
# In a function the moment the function sees return in execution it stops there. 
# Any line written after return is not valid
def classify_age(age):
    if age<13:
        return"Child"
    if age<20:
        return"Teenager"
    return "Adult",f"You are eligible to vote"

# When returning multiple values using the same return the values are packed into a tuple.

classify_age(12)
classify_age(15)
classify_age(30)
```

```python
def divide(a, b):
    """Divide two numbers.

    Args:
        a (float): The numerator.
        b (float): The denominator. Must not be zero.

    Returns:
        float: The result of a divided by b.

    Raises:
        ZeroDivisionError: If b is zero.

    Example:
        >>> divide(10, 2)
        5.0
    """
    return a / b

print(divide(10, 2))
```

# Numpy style documentation

```python
def convert_temp(celsius):
    """convert celsius to fahrenhiet.
    
    parameters
    ---------------
    celsius:float
            Temperature in celsius.

    Returns
    --------
    float
        temperature in fahrenhiet.
    """
    return (celsius*9/5)+32
```

```python
def calculate_discount(price, discount_percent):
    """
    Calculate the final price after applying a discount.

    Parameters
    ----------
    price : float
        The original price before discount.
    discount_percent : float
        The discount percentage to apply.

    Returns
    -------
    float
        The final price after the discount is applied.
    """
    return price - (price * discount_percent / 100)

print(calculate_discount(500, 10))
```

```python
def apply_tax(price, tax_rate=0.18):
    """
    Add tax to a given price.

    Parameters
    ----------
    price : float
        The pre-tax price.
    tax_rate : float, optional
        The tax rate to apply, by default 0.18.

    Returns
    -------
    float
        The price including tax.
    """
    return price + (price * tax_rate)

print(apply_tax(1000))
```
