# if statement

# if-elif chain

```python
grade = int(input("Enter the student's grade"))

if(grade>=90):
    print("A grade")
elif(grade>=80):
    print("B grade")
elif(grade>=70):
    print("C grade")
elif(grade>=60):
    print("D grade")
else:
    print("Fail")
```

# Truthy and Falsy

| Falsy values (the complete list) | Truthy = everything else                        |
|--------------------------------------|--------------------------------------------------------|
| 0, 0.0                                 | Any non-zero number                                       |
| "" (empty string)                       | Any non-empty string, even " " (just a space)               |
| [] (empty list)                         | Any non-empty list, even [0] or [False]                       |
| {} (empty dict)                          | Any non-empty dict                                              |
| set() (empty set)                         | Any non-empty set                                                 |
| () (empty tuple)                           | Any non-empty tuple                                                 |
| None                                        | —                                                                      |
| False                                        | True                                                                     |

```python
# if is only checking whether the variable has value or not
# Any value in the variable apart from the falsy values is considered as True
# The moment if see True it runs the statement inside the if condition

age =" "
if age:
    print("This person can enter the movie theater")
```

# match-case
# switch case in python

```python
# compares an exact match
# Used when matching one value against multiple patterns
# Runs which block matches first

day = input("Enter a day(e.g. Mon,Fri)")

match day:
    case "Mon":
        print("Start of the week")
    case "Fri":
        print("Almost weekend")
    case _:    # default case
        print("Regular day")
```

```python
grade = input("Enter the student's grade")

match grade:
    case "A" | "B":
        print("Good Performance")
    case "C" | "D":
        print("Needs Improvement")
    case _:
        print("Invalid grade")
```

```python
grade = input("Enter the student's grade")

if grade == "A" or "B":
    print("Good Performance")
elif grade == "C" or "D":
    print("Needs Improvement")
else:
    print("Invalid grade")
```
