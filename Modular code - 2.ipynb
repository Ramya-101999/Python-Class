# Flexible arguments  *args and **kwargs

```python
def add(*nums):
    return sum(nums)

add(7,8,9,10)
add(7,8)
add(7,8,10,17,20,15,16)
```

```python
def info(**details):
    print(details)

info(name="Sam",age=24,salary=56000,city="Mumbai")
```

```python
def info(name,*bill_amount):
    return f"The customer's name is: {name} and their total bill is:",sum(bill_amount)

info("Sameera",70,60,80,90)
```

```python
def info(name,*bill_amount,**details):
    return f"The customer's name is: {name} and their total bill is:",sum(bill_amount),f"details:{details}"

info("Sameera",70,60,80,90,city="Mumbai",table_no=67,items=6)
```

| Symbol      | In a function DEFINITION                     | In a function CALL                          |
|-----------------|----------------------------------------------------|----------------------------------------------------|
| *args              | Collects extra positional args into a tuple            | Spreads a list/tuple's items into separate positional args |
| **kwargs             | Collects extra keyword args into a dict                  | Spreads a dict's key-value pairs into keyword args            |
| Full valid order        | positional, *args, keyword-only (with defaults), **kwargs | —
