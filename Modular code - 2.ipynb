# Flexible arguments  *args and **kwargs

```python
def add(*nums):
    return sum(nums)

add(7,8,9,10)
add(7,8)
add(7,8,10,17,20,15,16)
```

```python
def info(**details):
    print(details)

info(name="Sam",age=24,salary=56000,city="Mumbai")
```

```python
def info(name,*bill_amount):
    return f"The customer's name is: {name} and their total bill is:",sum(bill_amount)

info("Sameera",70,60,80,90)
```

```python
def info(name,*bill_amount,**details):
    return f"The customer's name is: {name} and their total bill is:",sum(bill_amount),f"details:{details}"

info("Sameera",70,60,80,90,city="Mumbai",table_no=67,items=6)
```

| Symbol      | In a function DEFINITION                     | In a function CALL                          |
|-----------------|----------------------------------------------------|----------------------------------------------------|
| *args              | Collects extra positional args into a tuple            | Spreads a list/tuple's items into separate positional args |
| **kwargs             | Collects extra keyword args into a dict                  | Spreads a dict's key-value pairs into keyword args            |
| Full valid order        | positional, *args, keyword-only (with defaults), **kwargs | —                                           

```python
# Problem 1 

# Given imdb_scores = {"Whiplash": 8.5, "Cats": 2.8, "Parasite": 8.6, "Morbius": 3.2}
# Count how many movies have a score of 8.0 or above. Print the count.

imdb_scores = {"Whiplash": 8.5, "Cats": 2.8, "Parasite": 8.6, "Morbius": 3.2}

count = 0
for i,v in imdb_scores.items():
    if v>=8.0:
        count+=1
print(count)


# problem 2
# Given titles = ["Jaws", "The Godfather", "Her", "The Shawshank Redemption", "Up"]
# Build a list of only the titles longer than 10 characters. Print the list.

titles = ["Jaws", "The Godfather", "Her", "The Shawshank Redemption", "Up"]
title_long = []

for i in titles:
    if len(i)>10:
        title_long.append(i)
print(title_long)


# Problem 3

# Given imdb_scores = {"Whiplash": 8.5, "Cats": 2.8, "Parasite": 8.6, "Morbius": 3.2,
#                       "Oppenheimer": 8.3, "Madame Web": 3.7}
# Loop through and build TWO separate lists: "strong_movies" (score >= 8.0)
# and "weak_movies" (score < 4.0) — some movies will fall into neither.
# Print both lists, then print a one-line summary:
# "<X> strong, <Y> weak, out of <total> total"

imdb_scores = {"Whiplash": 8.5, "Cats": 2.8, "Parasite": 8.6, "Morbius": 3.2,
                       "Oppenheimer": 8.3, "Madame Web": 3.7}

strong_movies = []
weak_movies = []

for i,v in imdb_scores.items():
    if v >= 8.0:
        strong_movies.append(i)
    elif v < 4.0:
        weak_movies.append(i)

print("Strong movies:", strong_movies)
print("Weak movies:", weak_movies)

print(f"{len(strong_movies)} strong, {len(weak_movies)} weak, out of {len(imdb_scores)} total")
```

# Lambda Function 

Anonymous for one liner 

syntax - lambda parameters:expression

```python
# score > 7.0 then "Hit" else its a "Flop"

def rate(score):
    if score>7.0:
        return "Hit"
    else:
        return "Flop"
    
print(rate(8.0))
print(rate(6.0))

rate1 = lambda score1:"Hit" if score1>7.0 else "Flop"    # lambda method to write the same function in one line
print(rate1(8.0))
print(rate1(6.0))
```

```python
# price, pct

def apply_discount(price,pct):
    final_amount = price - (price*pct/100)
    return final_amount
print(apply_discount(6000,5))

apply_discount1 = lambda price1,pct1=10 : price1-(price1*pct1/100)
print(apply_discount1(6000,5))

print(apply_discount1(6000))
```

```python
# check if a genre is a commonly blockbuster genre

# genre-action, sci-fi, superhero

def blockbuster(genre):
    if genre in ["Action","sci-fi","superhero"]:
        return "Blockbuster"
    else:
        return "Not a Blockbuster genre"
    
print(blockbuster("Comedy"))
print(blockbuster("Action"))


blockbuster1 = lambda genre1:"Blockbuster" if genre1 in ["Action","sci-fi","superhero"] else "Not a blocbuster genre"
print(blockbuster1("Comedy"))
print(blockbuster1("Action"))
```

# lambda using nested ternary

when multiple conditions are to be checked we can use nested ternary

```python
classify = lambda score:"Excellent" if score>=8.5 else ("Good" if score>=6 else "poor")

print(classify(9.0))
print(classify(7.0))
print(classify(3.0))
```

```python
# lambda which returns mulyiple values at once

minmax = lambda num:(min(num),max(num))

print(minmax([7,8,9,10]))
```

# Functional Tools - map, filter and reduce

.map () - transform every item

syntax - list(map(function,iterable))

```python
titles = ["dune","oppenheimer","barbie"]
capitalized = list(map(str.title,titles))
print(capitalized)
```

```python
# mapping with multiple iterables 

prices = [1000,2000,3000]
discounts = [10,5,20]

final_prices = list(map(lambda p,d: p-(p*d/100),prices,discounts))
print(final_prices)


sq = [2,2,2]

square = list(map(lambda p,s: p**s,prices,sq))
print(square)

sq = [2,2,2]

square = list(map(lambda p: p**2,prices))
print(square)
```

# filter function

syntax -  list(filter(function,iterable))

```python
scores = [8.5,4.2,7.9,3.1,9.0]

hits = list(filter(lambda s: s>6, scores))
print(hits)

# filter out falsy values using None

mix = [0,1,"","hello",None,[],[1,2],False,True]

truthy = list(filter(None,mix))
print(truthy)
```
