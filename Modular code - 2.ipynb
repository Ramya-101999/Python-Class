# Flexible arguments  *args and **kwargs

```python
def add(*nums):
    return sum(nums)

add(7,8,9,10)
add(7,8)
add(7,8,10,17,20,15,16)
```

```python
def info(**details):
    print(details)

info(name="Sam",age=24,salary=56000,city="Mumbai")
```

```python
def info(name,*bill_amount):
    return f"The customer's name is: {name} and their total bill is:",sum(bill_amount)

info("Sameera",70,60,80,90)
```

```python
def info(name,*bill_amount,**details):
    return f"The customer's name is: {name} and their total bill is:",sum(bill_amount),f"details:{details}"

info("Sameera",70,60,80,90,city="Mumbai",table_no=67,items=6)
```

| Symbol      | In a function DEFINITION                     | In a function CALL                          |
|-----------------|----------------------------------------------------|----------------------------------------------------|
| *args              | Collects extra positional args into a tuple            | Spreads a list/tuple's items into separate positional args |
| **kwargs             | Collects extra keyword args into a dict                  | Spreads a dict's key-value pairs into keyword args            |
| Full valid order        | positional, *args, keyword-only (with defaults), **kwargs | —

runtimes = {"Parasite": 132,"The Godfather": 175,"Whiplash": 106, "Her": 126}
 loop through and track which movie has the SHORTEST runtime,printing "Shortest film: <title> (<minutes> min)".

```python
runtimes = {"Parasite": 132,"The Godfather": 175,"Whiplash": 106, "Her": 126}
shortest_runtime = 1000
for k,v in runtimes.items():
    if v<shortest_runtime:
        shortest_runtime = v
        shortest_film = k
print(f"shortest film:{shortest_film} ({shortest_runtime} min)")
```

genre_watched = ["Action", "Drama", "Action", "Comedy", "Drama", "Action"]
#Loop through and build a list if a genre appears more than once
#Print the final tally.

```python
genre_watched = ["Action", "Drama", "Action", "Comedy", "Drama", "Action"]

tally = []

for genre in genre_watched:
    if genre_watched.count(genre) > 1 and genre not in tally:
        tally.append(genre)

print(tally)
```

# H3 (Filtering a list of dicts with two combined conditions):
catalog = [{"title": "Dune 2", "genre": "Sci-Fi", "score": 8.5},
           {"title": "Barbie", "genre": "Comedy", "score": 6.9},
           {"title": "Interstellar", "genre": "Sci-Fi", "score": 8.7},
           {"title": "Anyone But You", "genre": "Comedy", "score": 6.0}]
Build a list of titles that are BOTH genre "Sci-Fi" AND scored 8.0 or above.

```python
catalog = [{"title": "Dune 2", "genre": "Sci-Fi", "score": 8.5},
           {"title": "Barbie", "genre": "Comedy", "score": 6.9},
           {"title": "Interstellar", "genre": "Sci-Fi", "score": 8.7},
           {"title": "Anyone But You", "genre": "Comedy", "score": 6.0}]

titles = []

for movie in catalog:
    if movie["genre"] == "Sci-Fi" and movie["score"] >= 8.0:
        titles.append(movie["title"])

print(titles)
```

 Q1 (List of dicts, each with a list inside):
 Given movies = [
    {"title": "Inception", "languages": ["English", "French"]},
     {"title": "RRR", "languages": ["Telugu", "Hindi", "Tamil"]}
 ]
Print "<title> is available in <language>" for every language of every movie.

```python
# Print "<title> is available in <language>" for every language of every movie.
movies = [
    {"title": "Inception", "languages": ["English", "French"]},
    {"title": "RRR", "languages": ["Telugu", "Hindi", "Tamil"]}
]

for movie in movies:
    for language in movie["languages"]:
        print(f"{movie['title']} is available in {language}")
```

 Q2 (Simple grid of showtimes):
 Given schedule = [["9AM", "1PM"], ["11AM", "3PM", "7PM"]]
 Print each day's showtimes on one line, space-separated.

```python
schedule = [["9AM", "1PM"], ["11AM", "3PM", "7PM"]]

for day in schedule:
    for time in day:
        print(time, end=" ")
    print()
```

Q3 (Star rating display):
 Given movie_stars = [("Inception", 5), ("Cats", 2)]
 For each movie, print that many "*" characters followed by the title,
 e.g. "*** Inception"

```python
movie_stars = [("Inception", 5), ("Cats", 2)]

for title, stars in movie_stars:
    for stars in range(stars):
        print("*",end=" ")
    print(f"{title}")
    print()
```
