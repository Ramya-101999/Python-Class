# Loops and Iteration

# For Loop

 For loop - iterating over iterables
 [4,5,6,7,9],(2,3,4,6),{1,2,3},{"k1":"1","k2":"2","k3":"3"},"hello","range"

 For loop is used to run the same code more than once or for each item in a collection

```python
# syntax - 

# for item in iterable:
#   repeatable code
sum = 0
l1 = [1,2,3,4,5]

for item in l1:
    sum += item
print(sum)
```

```python
wards = ["ICU","General","Cardiology"]
for w in wards:
    print(w)

# In the syntax of for loop item is a temporary variable that hols one value at a time
# The value is re-assigned everytime python goes into the for loop 
```

```python
# Looping over a string

word = "Eucalyptus"    # even if you give space, the space is also taken as character
for w in word:
    print(w)
```

```python
# Looping over a tuple and set

t1 = ("Hi","Welcome","to","our","home")
for t in t1:
    print(t)
print()

s1 = {"hi","welcome","to","our","store","to","store"}
for s in s1:
    print(s)
```

```python
# Looping over dictionary
# In a dictionary by default only keys are printed

d1 = {"mango":"yellow","banana":"yellow","pear":"green","watermelon":"green"}
for k in d1:
    print(k)

for k in d1.keys():
    print(k)

for k in d1.values():
    print(k)

for k,v in d1.items():
    print(k,":",v)
```

```python
# range function

for i in range(5):
    print(i)
```

```python
# start,stop and step in range
for i in range(1,7):
    print(i)

for i in range(1,7,2):  # step every 2nd element
    print(i)
```

| Iterable | What each loop step gives you                     |
|----------|--------------------------------------------------------|
| list       | Each item, in order                                       |
| string      | Each character, in order                                    |
| dict         | Each KEY by default; use .items() for (key, value) pairs      |
| set           | Each unique item, in no guaranteed order                        |
| tuple          | Each item, in order                                                |
| range(n)        | Numbers 0 to n-1                                                     |

# While Loop

while loops : infinite loops and exit conditions

A while loop keeps running as long as it's condition stay true.
Unlike for it doe not know when to stop automatically.

Syntax - 
while condition:
    code to run

```python
count = 1
while count<=5:
    print(count)
    count+=1
```

```python
count = True
while count<=5:
    print(count)
    count+=1
```

```python
# infinite loop
count = True
while count:
    print(count)
    count+=1
```

```python
# exit condition - (control keywords - break,continue,pass)
# break - as soon as it sees break it exits the loop immediately

total = 0
while True:
    entry = input("Enter a bill amount(or 'done' to stop):")
    if entry == 'done':
        break
    total += int(entry)
print("total:",total)
```

```python
# break

numbers = [17,18,19,20,26,22,16]
for n in numbers:
    if n==16:
        print("Found 16!, Stopping my search")
        break
    print("Checking:",n)
```

```python
# continue - skip just this one iteration, keep looping
numbers = [17,18,19,16,20,26,22,16]
for n in numbers:
    if n%2!= 0:
        continue
    print("Even Number:",n)
```

```python
# pass - placeholder, do nothing

status = "pending"
if status == "approved":
    print("Approved")
elif status == "pending":
    pass
else:
    print("Rejected")
print("Done checking status")
```

```python
# break and continue inside while loop

count = 0
while True:
    count +=1
    if count == 3:
        break
    print("count is:", count)
```

```python
# continue with while loop

n = 0
while n<6:
    n+=1
    if n%2 == 0:
        continue
    print("Odd Numbers:",n)
```
