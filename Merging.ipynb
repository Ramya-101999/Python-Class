```python
import pandas as pd

df = pd.read_csv("all_athlete_games.csv", low_memory=False)
sample = df[(df["Year"]==2020) & (df["Medal"].notna())][["Name","NOC","Sport","Medal"]].head(50).reset_index(drop=True)
print(sample)
```

# Concatenation

```python
# Concatenation: Vertical (axis=0) vs Horizontal (axis=1)

# syntax - pd.concat([df1,df2,.....],axis=0,ignore_index=False)

# axis=0 (default) - stacking rows on top of each other

batch1 = sample.iloc[0:3].reset_index(drop=True)
batch2= sample.iloc[7:11].reset_index(drop=True)

combined = pd.concat([batch1,batch2],ignore_index=True)
print(combined)
```

```python
# Stacking columns side by side, axis=1

extra_info = pd.DataFrame({
    "Height_cm":[186,178,165],
    "Weight_kg":[88,65,55]
})

horiz = pd.concat([batch1,extra_info],axis=1)
print(horiz)
```

```python
# Merging - pd.merge()

# syntax - pd.merge(left,right,on=None,left_on=None,right_on=None,how="inner")

# Unlike concat(axis=1) which lines up by position, merge() lines up rows based on matching
# values in a key column - much safer and more common for combining real related datasets

# Basic merge

noc_lookup = pd.DataFrame({
    "NOC": ["FRA","USA","EGY","BEL","ZZZ"],
    "Country": ["France","United States","Egypt","Belgium","Nowhereland"],
    "Continent": ["Europe","North America","Africa","Europe","Unknown"]
})
print(noc_lookup)
```

```python
merged = pd.merge(sample,noc_lookup,on="NOC")
print(merged[["Name","NOC","Country"]])
```

```python
# left_on, right_on

sample2 = pd.DataFrame({
    "Name":["ABALO Luc","ABBOT Monica"],
    "NOC":["FRA","USA"]
})

lookup2 = pd.DataFrame({
    "Country_code":["FRA","USA"],
    "Country":["France","United States"]
})

merged2 = pd.merge(sample2,lookup2,left_on="NOC",right_on="Country_code")
print(merged2)
```
